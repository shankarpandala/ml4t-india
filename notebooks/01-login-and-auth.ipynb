## 1. Build the login URL

Kite requires the user to visit a login URL, authenticate, and return with a `request_token` query parameter.

```python
import getpass
from ml4t.india.kite.auth import generate_session

api_key    = getpass.getpass("Kite API key: ")
api_secret = getpass.getpass("Kite API secret: ")
request_token = getpass.getpass("Request token (from redirect URL): ")

record = generate_session(
    api_key=api_key,
    api_secret=api_secret,
    request_token=request_token,
)
print(record.user_id, record.access_token[:4] + "***REDACTED***")
```

## 2. Exchange request_token for access_token

After clicking the URL above and logging in to Zerodha, your browser is redirected to
`http://127.0.0.1:5678?request_token=XXX&status=success`.

Copy the `request_token` value from the browser address bar and paste it into `generate_session` below.
